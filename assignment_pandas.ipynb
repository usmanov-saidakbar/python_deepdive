```python
# loading the dataset
data = pd.read_csv("./data/adult.data.csv")
data.head()
```

**1. Calculate how many men and women (attribute *sex*) are represented in this dataset.**

**2. What is the average age of men (attribute *age*) across the entire dataset?**

**3. What is the proportion of United States citizens (*native-country* attribute)?**

**4-5. Calculate the mean and standard deviation of the age of those who earn more than 50K per year (the *salary* attribute) and those who earn less than 50K per year**

```python
less_age_mean=data[data['salary']=='<=50K']['age'].mean()
less_age_std=data[data['salary']=='<=50K']['age'].std()

large_age_mean=data[data['salary']=='>50K']['age'].mean()
large_age_std=data[data['salary']=='>50K']['age'].std()


print(f'less than 50k: {int(less_age_mean),int(less_age_std)}')
print(f'larger than 50k: {int(large_age_mean),int(large_age_std)}')
```

**6. Is it true that people who earn more than 50k have at least a higher education? (*education indicator – Bachelors, Prof-school, Assoc-acdm, Assoc-voc, Masters or Doctorate*)**

```python
high_edu_count=data[data['salary']=='>50K']['education'].isin(['Bachelors','Prof-school','Assocacdm','Assoc-voc','Masters','Doctorate']).count()
total_high_50_count=len(data[data['salary']=='>50K'])

# i got 100 percent which means all of the people has higher education 
print(f'percent of people who has more salary than 50k and has high edu {100*high_edu_count/total_high_50_count}')
```

**7. Find age statistics for each race (*race* attribute) and each gender. Use *groupby* and *describe*. Find the maximum age of males of the *Asian-Pac-Islander* race.**

```python
data[(data["race"] == "Asian-Pac-Islander") & (data["sex"] == "Male")]["age"].max()
```

**8. Among whom is the proportion of high-earning individuals (>50K) higher: married or single men (*marital-status* indicator)? We consider those whose *marital-status* begins with *Married* (Married-civ-spouse, Married-spouse-absent, or Married-AF-spouse) to be married; the rest are considered single.**

```python

married_counts = len(data[(data['salary']=='>50K') & (data['sex']=='Male') & (data['marital-status'].str.startswith('Married',na=False))])
# we just use ~ before marital-staus filter to make it notstartwith
single_count = len(data[(data['salary']=='>50K') & (data['sex']=='Male') & (~data['marital-status'].str.startswith('Married',na=False))])
total = len(data[(data['salary']=='>50K') & (data['sex']=='Male')])

print(f'single who earns higher than 50k is about {100*single_count//total} % and married {100*married_counts//total} %')

```

**9. What is the maximum number of hours a person works per week (hours-per-week indicator)? How many people work that many hours, and what percentage of them earn a lot?**

```python
max_work_hour=data['hours-per-week'].max()

how_many_work_many = len(data[data['hours-per-week']>=max_work_hour-1])

count_who_earn_alot= len(data[(data['hours-per-week']>=max_work_hour-1) & (data['salary']=='>50K')])

count_who_earn_alot_percentage = 100*count_who_earn_alot/how_many_work_many

print(f'max work hour is {max_work_hour} and {how_many_work_many} people work such, \n and {count_who_earn_alot} people earn alot and they are {round(count_who_earn_alot_percentage,2)}% of them')
```

**10. Calculate the average working hours (*hours-per-week*) of low and high earners (*salary*) for each country (*native-country*).**

```python
#unstack makes it easier to read by making it colum
data.groupby(['native-country', 'salary'])['hours-per-week'].mean().unstack().sort_values('<=50K',ascending=False)
```

**11.Group people into age groups *young*, *adult*, *retiree*, where:**
* *young* corresponds to 16-35 years old
* *adult* - 35-70 years old
* *retiree* - 70-100 years old

**Enter the name of the appropriate age group for each person in the new AgeGroup column.**

```python
data['AgeGroup'] = pd.cut(
    data['age'],
    bins=[16,35,70,100],
    labels=['young','adult','retiree']
)
```

**12-13. Determine the number of people earning >50K in each age group (AgeGroup column), and also output the name of the age group in which people most often earn more than 50K (>50K)**

```python
data_temp = data[data['salary']=='>50K']
data_temp.groupby(['AgeGroup','salary']).size()
```

**14. Group people by employment type (the occupation column) and determine the number of people in each group. Then write a filter function, filter_func, that will return only those groups in which the average age (the age column) is 40 or less and in which all employees work more than 5 hours per week (the hours-per-week column).**

```python
#size of each occupation 
print(data.groupby('occupation').size())

def filter_func(df):
    return df['age'].mean()<=40 and (df['hours-per-week']>5).all()

data.groupby('occupation').filter(filter_func)
```
